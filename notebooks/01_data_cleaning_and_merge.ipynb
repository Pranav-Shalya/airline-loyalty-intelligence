# Airline Loyalty Intelligence - 01 Data Cleaning and Ground Truth Definition

## Goal & Objective
The primary goal of this notebook is to ingest, clean, and consolidate raw airline customer loyalty data into a standardized Analytical Base Table (ABT) for downstream feature engineering and churn modeling.

### Key Steps:
1. **Load Raw Data**: Ingest the 4 raw CSV datasets from `../data/raw/`:
   - `Airline Loyalty Data Dictionary.csv`: Metadata and column descriptions.
   - `Calendar.csv`: Date dimensions and fiscal period mappings.
   - `Customer Flight Activity.csv`: Granular monthly customer flight behaviors (flights booked, points accumulated/redeemed, dollar cost).
   - `Customer Loyalty History.csv`: Customer demographics, enrollment dates, loyalty tiers, cancellation/churn status, and CLV metrics.
2. **Data Cleaning & Type Standardization**: Parse datetime fields, clean categorical formats, validate numeric constraints, and handle missing values.
3. **Ground Truth Churn Definition**: Formulate explicit churn and cancellation indicators and temporal cutoffs.
4. **Merge & Validation**: Perform relational joins on `Loyalty Number` across calendar periods and save processed tables to `../data/processed/`.

```python
from pathlib import Path
import numpy as np
import pandas as pd

# Set display preferences
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Setup data directory paths
DATA_RAW_DIR = Path('../data/raw')
DATA_PROCESSED_DIR = Path('../data/processed')
DATA_PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Raw CSV file paths
PATH_DATA_DICT = DATA_RAW_DIR / 'Airline Loyalty Data Dictionary.csv'
PATH_CALENDAR = DATA_RAW_DIR / 'Calendar.csv'
PATH_FLIGHT_ACTIVITY = DATA_RAW_DIR / 'Customer Flight Activity.csv'
PATH_LOYALTY_HISTORY = DATA_RAW_DIR / 'Customer Loyalty History.csv'

# Load raw datasets
df_data_dict = pd.read_csv(PATH_DATA_DICT)
df_calendar = pd.read_csv(PATH_CALENDAR)
df_flight_activity = pd.read_csv(PATH_FLIGHT_ACTIVITY)
df_loyalty_history = pd.read_csv(PATH_LOYALTY_HISTORY)

# Display summary shapes and preview
print(f"Data Dictionary shape:    {df_data_dict.shape}")
print(f"Calendar shape:           {df_calendar.shape}")
print(f"Flight Activity shape:    {df_flight_activity.shape}")
print(f"Loyalty History shape:    {df_loyalty_history.shape}")

display(df_loyalty_history.head())
```
